# Layer 2: Intermediate ESI Random Forest & Hierarchical Cascade with Combined Features (`models/rf_combined_extreme.ipynb`)

This notebook trains the **Layer 2 Tuned Combined Feature Model** and evaluates the **2-Layer Combined Hierarchical Cascade System**:
- **Combined Feature Predictors (16 Total)**:
  - **6 Raw Features**: `age`, `gender`, `triage_vital_hr`, `triage_vital_sbp`, `triage_vital_rr`, `triage_vital_o2`.
  - **10 Feature Engineered Flags**: `is_dyspnea_total`, `is_dyspnea_moderate`, `is_bradypnea`, `is_tachypnea`, `is_hypotension`, `is_hypertension`, `is_bradycardia_total`, `is_bradycardia_moderate`, `is_tachycardia_total`, `is_tachycardia_moderate`.
- **Hyperparameter Tuning**: Grid search tuning over `mtry`, `min.node.size`, and `splitrule` on the validation set.
- **Layer 2 Model**: Random Forest (`ranger`) trained on ESI 2, ESI 3, and ESI 4, **excluding rows containing ESI 1 and 5**.
- **Hierarchical Cascade Flow**:
  - Layer 1 (Combined Feature Logistic Regression) evaluates sample: returns ESI 1, ESI 5, or 'neither'.
  - If 'neither', Layer 2 (Combined Feature Tuned Random Forest) evaluates sample: decides ESI 2, ESI 3, or ESI 4.
- **Scoring Metrics**: **Accuracy**, **Multi-Class ROC-AUC**, and **Log Loss**.

```python
%%R
# ---------------------------------------------------------
# Step 1: Load Required Libraries & Parse Configuration JSON
# ---------------------------------------------------------
library(jsonlite)
library(caret)
library(ranger)
library(nnet)
library(dplyr)
library(ggplot2)
library(pROC)

config_path <- "../config/triage_conf.json"
if (!file.exists(config_path)) {
  config_path <- "config/triage_conf.json"
}

config <- fromJSON(config_path)

cat("=== Configuration Loaded from config/triage_conf.json ===\n")
cat("Data Source Path:", config$path$data_source, "\n")
cat("Target Column:   ", config$classes$target_col, "\n")
cat("Test Size:       ", config$training$test_size, "\n")
cat("Val Size:        ", config$training$val_size, "\n")
cat("Random State:    ", config$training$random_state, "\n")
```

```python
%%R
# ---------------------------------------------------------
# Step 2: Load Data, Compute Combined Features, & Filter Out ESI 1 & 5
# ---------------------------------------------------------
set.seed(config$training$random_state)

data_file <- config$path$data_source
if (!file.exists(data_file) && file.exists(paste0("../", data_file))) {
  data_file <- paste0("../", data_file)
}

cat("Loading dataset from:", data_file, "...\n")

data_env <- new.env()
load(data_file, envir = data_env)

df_names <- ls(data_env)[sapply(ls(data_env), function(x) is.data.frame(get(x, envir = data_env)))]
df_sizes <- sapply(df_names, function(x) nrow(get(x, envir = data_env)))
data_obj_name <- df_names[which.max(df_sizes)]
cat(sprintf("Selected main dataset object: '%s' (%d rows)\n", data_obj_name, max(df_sizes)))

raw_df <- get(data_obj_name, envir = data_env)
target_col <- config$classes$target_col

gender_vec <- if ("gender" %in% names(raw_df)) ifelse(as.character(raw_df$gender) == "Male", 1, 0) else 0

# Construct Combined Feature Set: 6 Raw Features + 10 Feature Engineered Flags
df_combined <- data.frame(
  # Raw Features
  age                     = raw_df$age,
  gender                  = gender_vec,
  triage_vital_hr         = raw_df$triage_vital_hr,
  triage_vital_sbp        = raw_df$triage_vital_sbp,
  triage_vital_rr         = raw_df$triage_vital_rr,
  triage_vital_o2         = raw_df$triage_vital_o2,
  
  # Feature Engineered Flags
  is_dyspnea_total        = ifelse(!is.na(raw_df$triage_vital_o2) & raw_df$triage_vital_o2 < 90, 1, 0),
  is_dyspnea_moderate     = ifelse(!is.na(raw_df$triage_vital_o2) & raw_df$triage_vital_o2 >= 90 & raw_df$triage_vital_o2 < 94, 1, 0),
  is_bradypnea            = ifelse(!is.na(raw_df$triage_vital_rr) & raw_df$triage_vital_rr < 10, 1, 0),
  is_tachypnea            = ifelse(!is.na(raw_df$triage_vital_rr) & raw_df$triage_vital_rr > 30, 1, 0),
  is_hypotension          = ifelse(!is.na(raw_df$triage_vital_sbp) & raw_df$triage_vital_sbp <= 90, 1, 0),
  is_hypertension         = ifelse(!is.na(raw_df$triage_vital_sbp) & raw_df$triage_vital_sbp > 220, 1, 0),
  is_bradycardia_total    = ifelse(!is.na(raw_df$triage_vital_hr) & raw_df$triage_vital_hr < 40, 1, 0),
  is_bradycardia_moderate = ifelse(!is.na(raw_df$triage_vital_hr) & raw_df$triage_vital_hr >= 40 & raw_df$triage_vital_hr < 60, 1, 0),
  is_tachycardia_total    = ifelse(!is.na(raw_df$triage_vital_hr) & raw_df$triage_vital_hr > 150, 1, 0),
  is_tachycardia_moderate = ifelse(!is.na(raw_df$triage_vital_hr) & raw_df$triage_vital_hr > 100 & raw_df$triage_vital_hr <= 150, 1, 0)
)
df_combined[[target_col]] <- raw_df[[target_col]]

# Filter dataset for Layer 2: REMOVE ESI 1 and ESI 5
raw_esi_char <- as.character(df_combined[[target_col]])
df_layer2 <- df_combined[raw_esi_char %in% c("2", "3", "4"), ]
df_layer2[[target_col]] <- factor(as.character(df_layer2[[target_col]]), levels = c("2", "3", "4"))

if (any(is.na(df_layer2))) {
  df_layer2 <- na.omit(df_layer2)
}

cat(sprintf("Layer 2 Filtered Dataset Ready (No ESI 1 & 5): %d rows x %d cols\n", nrow(df_layer2), ncol(df_layer2)))
cat("Layer 2 Target Distribution:\n")
print(table(df_layer2[[target_col]]))
```

```python
%%R
# ---------------------------------------------------------
# Step 3: Stratified Data Partitioning for Layer 2
# ---------------------------------------------------------
set.seed(config$training$random_state)

test_size <- config$training$test_size
val_size  <- config$training$val_size

in_train_val <- createDataPartition(df_layer2[[target_col]], p = 1 - test_size, list = FALSE)
train_val_df <- df_layer2[in_train_val, ]
test_df_l2   <- df_layer2[-in_train_val, ]

rel_val_size <- val_size / (1 - test_size)
in_train    <- createDataPartition(train_val_df[[target_col]], p = 1 - rel_val_size, list = FALSE)
train_df_l2 <- train_val_df[in_train, ]
val_df_l2   <- train_val_df[-in_train, ]

cat(sprintf("Layer 2 Partitions:\n  Train: %d rows\n  Val:   %d rows\n  Test:  %d rows\n",
            nrow(train_df_l2), nrow(val_df_l2), nrow(test_df_l2)))
```

```python
%%R
# ---------------------------------------------------------
# Step 3B: Hyperparameter Tuning Grid Search for Layer 2 Random Forest
# ---------------------------------------------------------
set.seed(config$training$random_state)

cat("=== Starting Layer 2 Combined Random Forest Hyperparameter Grid Search ===\n")

param_grid <- expand.grid(
  mtry = c(3, 4, 6, 8),
  min_node_size = c(1, 5, 10, 20),
  splitrule = c("gini", "extratrees"),
  stringsAsFactors = FALSE
)

results_list <- list()
formula_rf <- as.formula(paste(target_col, "~ ."))
target_classes_l2 <- levels(val_df_l2[[target_col]])

for (i in 1:nrow(param_grid)) {
  cur_mtry      <- param_grid$mtry[i]
  cur_node_size <- param_grid$min_node_size[i]
  cur_splitrule <- param_grid$splitrule[i]
  
  fit_tmp <- ranger(
    formula = formula_rf,
    data = train_df_l2,
    num.trees = 150,
    mtry = cur_mtry,
    min.node.size = cur_node_size,
    splitrule = cur_splitrule,
    probability = TRUE,
    seed = config$training$random_state
  )
  
  probs_val <- predict(fit_tmp, data = val_df_l2)$predictions
  max_idx   <- max.col(probs_val, ties.method = "first")
  preds_val <- factor(colnames(probs_val)[max_idx], levels = target_classes_l2)
  actuals_val <- factor(val_df_l2[[target_col]], levels = target_classes_l2)
  
  acc_val <- mean(preds_val == actuals_val)
  roc_val <- tryCatch(as.numeric(pROC::multiclass.roc(actuals_val, probs_val)$auc), error = function(e) NA)
  
  results_list[[i]] <- data.frame(
    mtry = cur_mtry,
    min_node_size = cur_node_size,
    splitrule = cur_splitrule,
    val_accuracy = acc_val,
    val_roc_auc = roc_val
  )
}

tuning_df <- do.call(rbind, results_list)
tuning_df <- tuning_df[order(-tuning_df$val_accuracy, -tuning_df$val_roc_auc), ]

cat("\n=== Top 5 Hyperparameter Configurations Found ===\n")
print(head(tuning_df, 5))

best_params <- tuning_df[1, ]
cat(sprintf("\nOptimal Hyperparameters Selected:\n  mtry: %d\n  min.node.size: %d\n  splitrule: %s\n  Validation Accuracy: %.4f\n",
            best_params$mtry, best_params$min_node_size, best_params$splitrule, best_params$val_accuracy))
```

```python
%%R
# ---------------------------------------------------------
# Step 4: Train Optimal Tuned Layer 2 Combined Feature Random Forest Model
# ---------------------------------------------------------
set.seed(config$training$random_state)

formula_rf <- as.formula(paste(target_col, "~ ."))

cat(sprintf("Training Tuned Layer 2 Combined Random Forest (mtry=%d, min.node.size=%d, splitrule='%s')...\n",
            best_params$mtry, best_params$min_node_size, best_params$splitrule))

rf_combined_layer2 <- ranger(
  formula = formula_rf,
  data = train_df_l2,
  num.trees = 300,
  mtry = best_params$mtry,
  min.node.size = best_params$min_node_size,
  splitrule = best_params$splitrule,
  importance = "impurity",
  probability = TRUE,
  seed = config$training$random_state
)

cat("Tuned Layer 2 Combined Random Forest training complete!\n")
print(rf_combined_layer2)
```

```python
%%R
# ---------------------------------------------------------
# Step 5: Standalone Tuned Layer 2 Benchmark (Accuracy, ROC-AUC, Log Loss)
# ---------------------------------------------------------

calc_log_loss <- function(actual_factor, prob_matrix, eps = 1e-15) {
  prob_matrix <- pmax(pmin(prob_matrix, 1 - eps), eps)
  prob_matrix <- prob_matrix / rowSums(prob_matrix)
  classes <- colnames(prob_matrix)
  N <- length(actual_factor)
  
  log_probs <- numeric(N)
  for (i in 1:N) {
    act_cls <- as.character(actual_factor[i])
    if (act_cls %in% classes) {
      log_probs[i] <- log(prob_matrix[i, act_cls])
    } else {
      log_probs[i] <- log(eps)
    }
  }
  return(-mean(log_probs))
}

evaluate_layer2_rf <- function(model, data, set_name, target_col) {
  prob_matrix <- predict(model, data = data)$predictions
  target_classes <- levels(data[[target_col]])
  
  max_idx <- max.col(prob_matrix, ties.method = "first")
  pred_factor <- factor(colnames(prob_matrix)[max_idx], levels = target_classes)
  actual_factor <- factor(data[[target_col]], levels = target_classes)
  
  cm <- confusionMatrix(pred_factor, actual_factor)
  acc <- as.numeric(cm$overall["Accuracy"])
  
  roc_auc <- tryCatch({
    as.numeric(pROC::multiclass.roc(actual_factor, prob_matrix)$auc)
  }, error = function(e) NA)
  
  log_loss <- calc_log_loss(actual_factor, prob_matrix)
  
  cat(sprintf("============================================================\n"))
  cat(sprintf("   STANDALONE TUNED LAYER 2 COMBINED RF - %s SET BENCHMARK\n", toupper(set_name)))
  cat(sprintf("============================================================\n"))
  cat(sprintf("  Accuracy   : %.4f (%.2f%%)\n", acc, acc * 100))
  cat(sprintf("  ROC-AUC    : %.4f\n", roc_auc))
  cat(sprintf("  Log Loss   : %.4f\n", log_loss))
  cat("\nConfusion Matrix:\n")
  print(cm$table)
  cat(sprintf("============================================================\n\n"))
}

# Standalone Layer 2 Combined Evaluations
evaluate_layer2_rf(rf_combined_layer2, val_df_l2, "Validation", target_col)
evaluate_layer2_rf(rf_combined_layer2, test_df_l2, "Test", target_col)
```

```python
%%R
# ---------------------------------------------------------
# Step 6: Full 2-Layer Combined Feature Hierarchical Cascade Evaluation
# ---------------------------------------------------------

# Load Layer 1 Combined Feature Model
lr_path <- file.path("../deploy", "lr_combined_extreme_model.rds")
if (!file.exists(lr_path)) lr_path <- file.path("deploy", "lr_combined_extreme_model.rds")

if (file.exists(lr_path)) {
  lr_obj <- readRDS(lr_path)
  cat("Successfully loaded Layer 1 Combined Feature LR model from:", lr_path, "\n")
} else {
  cat("Layer 1 Combined model not found at:", lr_path, ". Training quick inline Layer 1 Combined model...\n")
  df_l1 <- df_combined
  raw_esi_c <- as.character(df_l1[[target_col]])
  df_l1$target_layer1 <- factor(ifelse(raw_esi_c == "1", "1", ifelse(raw_esi_c == "5", "5", "neither")), levels = c("1", "5", "neither"))
  cont_cols <- c("age", "triage_vital_hr", "triage_vital_sbp", "triage_vital_rr", "triage_vital_o2")
  preproc_l1 <- preProcess(df_l1[, cont_cols, drop = FALSE], method = c("center", "scale"))
  df_l1[, cont_cols] <- predict(preproc_l1, df_l1[, cont_cols, drop = FALSE])
  formula_l1 <- as.formula(paste("target_layer1 ~", paste(setdiff(names(df_l1), c(target_col, "target_layer1")), collapse = " + ")))
  lr_mod_inline <- multinom(formula_l1, data = df_l1, trace = FALSE)
  lr_obj <- list(model = lr_mod_inline, preproc = preproc_l1)
}

# Predict Function for Combined Feature 2-Layer Cascade System
predict_hierarchical_cascade <- function(lr_obj, rf_model, newdata, target_col) {
  N <- nrow(newdata)
  target_classes_5 <- c("1", "2", "3", "4", "5")
  
  # 1. Scale continuous variables for Layer 1
  data_l1 <- newdata
  if (!is.null(lr_obj$preproc)) {
    cont_cols <- c("age", "triage_vital_hr", "triage_vital_sbp", "triage_vital_rr", "triage_vital_o2")
    data_l1[, cont_cols] <- predict(lr_obj$preproc, data_l1[, cont_cols, drop = FALSE])
  }
  
  # 2. Layer 1 Probabilities (P(1), P(5), P(neither))
  prob_l1 <- predict(lr_obj$model, newdata = data_l1, type = "probs")
  if (!is.matrix(prob_l1)) {
    prob_l1 <- matrix(prob_l1, nrow = N, ncol = 3, byrow = TRUE)
    colnames(prob_l1) <- c("1", "5", "neither")
  }
  
  # 3. Layer 2 Probabilities (P(2), P(3), P(4))
  prob_l2 <- predict(rf_model, data = newdata)$predictions
  
  # 4. Construct Unified 5-Class Probability Matrix
  prob_5class <- matrix(0, nrow = N, ncol = 5)
  colnames(prob_5class) <- target_classes_5
  
  prob_5class[, "1"] <- prob_l1[, "1"]
  prob_5class[, "5"] <- prob_l1[, "5"]
  
  p_neither <- prob_l1[, "neither"]
  prob_5class[, "2"] <- p_neither * prob_l2[, "2"]
  prob_5class[, "3"] <- p_neither * prob_l2[, "3"]
  prob_5class[, "4"] <- p_neither * prob_l2[, "4"]
  
  # Normalize probabilities per sample
  prob_5class <- prob_5class / rowSums(prob_5class)
  
  max_idx <- max.col(prob_5class, ties.method = "first")
  pred_classes <- factor(target_classes_5[max_idx], levels = target_classes_5)
  
  return(list(predictions = pred_classes, probabilities = prob_5class))
}

evaluate_hierarchical_cascade <- function(lr_obj, rf_model, test_data, set_name, target_col) {
  res <- predict_hierarchical_cascade(lr_obj, rf_model, test_data, target_col)
  target_classes_5 <- c("1", "2", "3", "4", "5")
  actual_factor <- factor(as.character(test_data[[target_col]]), levels = target_classes_5)
  
  cm <- confusionMatrix(res$predictions, actual_factor)
  acc <- as.numeric(cm$overall["Accuracy"])
  
  roc_auc <- tryCatch({
    as.numeric(pROC::multiclass.roc(actual_factor, res$probabilities)$auc)
  }, error = function(e) NA)
  
  log_loss <- calc_log_loss(actual_factor, res$probabilities)
  
  cat(sprintf("============================================================\n"))
  cat(sprintf("   2-LAYER COMBINED FEATURE CASCADE SYSTEM (%s SET)\n", toupper(set_name)))
  cat(sprintf("============================================================\n"))
  cat(sprintf("  Overall Accuracy : %.4f (%.2f%%)\n", acc, acc * 100))
  cat(sprintf("  Multi-Class ROC-AUC : %.4f\n", roc_auc))
  cat(sprintf("  Log Loss            : %.4f\n", log_loss))
  cat("\nFull 5-Class Confusion Matrix:\n")
  print(cm$table)
  cat(sprintf("============================================================\n\n"))
}

# Evaluate Hierarchical Cascade on Full Dataset Test Split
in_train_val_full <- createDataPartition(df_combined[[target_col]], p = 1 - test_size, list = FALSE)
test_df_full <- df_combined[-in_train_val_full, ]

evaluate_hierarchical_cascade(lr_obj, rf_combined_layer2, test_df_full, "Full Test Split", target_col)
```

```python
%%R
# ---------------------------------------------------------
# Step 7: Save Combined Feature Model Artifacts
# ---------------------------------------------------------
deploy_dir <- "../deploy"
if (!dir.exists(deploy_dir)) deploy_dir <- "deploy"
if (!dir.exists(deploy_dir)) dir.create(deploy_dir, recursive = TRUE)

saveRDS(rf_combined_layer2, file.path(deploy_dir, "rf_combined_extreme_model.rds"))
cat("Layer 2 Tuned Combined Feature Random Forest saved to:", file.path(deploy_dir, "rf_combined_extreme_model.rds"), "\n")

cascade_obj <- list(lr_model = lr_obj, rf_model = rf_combined_layer2, best_params = best_params)
class(cascade_obj) <- "hierarchical_triage_combined_cascade"
saveRDS(cascade_obj, file.path(deploy_dir, "hierarchical_combined_cascade_model.rds"))
cat("Complete Tuned Combined Feature Hierarchical Cascade System saved to:", file.path(deploy_dir, "hierarchical_combined_cascade_model.rds"), "\n")
```
